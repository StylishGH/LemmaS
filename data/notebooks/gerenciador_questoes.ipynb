```python
def remover_latex(texto):                                                                          
      if texto is None or pd.isna(texto):                                                            
          return ""                                                                                  
                                                                                                     
      texto = str(texto)                                                                             
                                                                                                     
      # Remove blocos matemáticos delimitados por LaTeX                                              
      padroes_matematicos = [                                                                        
          r"\\\[.*?\\\]",  # \[ ... \]                                                               
          r"\\\(.*?\\\)",  # \( ... \)                                                               
          r"\$\$.*?\$\$",  # $$ ... $$                                                               
          r"(?<!\\)\$.*?(?<!\\)\$",  # $ ... $                                                       
          r"\\begin\{(?:equation|align|gather|multline|displaymath)\*?\}.*?"                         
          r"\\end\{(?:equation|align|gather|multline|displaymath)\*?\}",                             
      ]                                                                                              
                                                                                                     
      for padrao in padroes_matematicos:                                                             
          texto = re.sub(padrao, " ", texto, flags=re.DOTALL)                                        
                                                                                                     
      # Limpa comandos LaTeX que sobrarem e organiza os espaços                                      
      texto = re.sub(r"\\[a-zA-Z]+\*?(?:\[[^\]]*\])?", " ", texto)                                   
      texto = re.sub(r"[{}]", " ", texto)                                                            
      texto = re.sub(r"\s+", " ", texto)                                                             
                                                                                                     
      return texto.strip()
```

```python
import re

MAPA_EXPANDIDO = {
    # Espanhol
    "spanish": "es", "español": "es", "espanol": "es", "castellano": "es",
    # Inglês
    "english": "en", "ingles": "en", "inglês": "en",
    # Português
    "portuguese": "pt", "português": "pt", "portugues": "pt",
    # Francês
    "french": "fr", "français": "fr", "francês": "fr",
    # Alemão
    "german": "de", "deutsch": "de",
    # Italiano
    "italian": "it", "italiano": "it",
    # Chinês
    "chinese": "zh", "chinese (traditional)": "zh", "chinese (simplified)": "zh", "mandarin": "zh",
    # Outros
    "russian": "ru", "mongolian": "mn", "romanian": "ro", "slovenian": "sl",
    "dutch": "nl", "korean": "ko", "turkish": "tr", "greek": "el",
    "ukrainian": "uk", "bulgarian": "bg", "croatian": "hr", "japanese": "ja",
    "czech": "cs", "hungarian": "hu", "polish": "pl", "vietnamese": "vi",
    "estonian": "et", "macedonian": "mk", "arabic": "ar", "hindi": "hi",
    "persian": "fa", "swedish": "sv", "finnish": "fi", "norwegian": "no",
    "catalan": "ca", "danish": "da"
}

def corrigir_idioma(row):
    # Normaliza o idioma reconhecido na etapa anterior
    valor = str(row["idioma_final"]).strip().lower() if pd.notna(row["idioma_final"]) else ""
    
    # 1. Se tem múltiplos nomes com vírgula ou barra (ex: "english, arabic", "macedonian, english")
    if re.search(r"[,;/]| and ", valor):
        detectado = detectar_idioma_seguro(row["enunciado_sem_latex"])
        if detectado:
            return detectado.split("-")[0]
            
    # 2. Se for um nome por extenso (ex: "español" -> "es", "chinese (traditional)" -> "zh")
    if valor in MAPA_EXPANDIDO:
        return MAPA_EXPANDIDO[valor]
        
    # 3. Se já for código de 2 letras limpo (ex: "en", "pt", "es")
    if len(valor) == 2 and valor.isalpha():
        return valor
        
    # 4. Se tiver hífen (ex: "zh-cn" -> "zh")
    if "-" in valor:
        return valor.split("-")[0]
        
    # 5. Se nada acima resolveu, tenta o detector no texto
    detectado = detectar_idioma_seguro(row["enunciado_sem_latex"])
    if detectado:
        return detectado.split("-")[0]
        
    return None

# Atualiza diretamente a coluna 'idioma_limpo' e 'idioma'
df["idioma_limpo"] = df.apply(corrigir_idioma, axis=1)
df["idioma"] = df["idioma_limpo"]

# Exibe o resultado da própria 'idioma_limpo'
print("🌍 Ranking Final de idioma_limpo:")
print(df["idioma_limpo"].value_counts(dropna=False).head(20))
```

## 8. Ingestão e Integração Canônica com Supabase (LEMMAS Core)

Nesta etapa, preparamos e enviamos os registros para o banco de dados oficial (Supabase PostgreSQL) preservando a integridade matemática e os novos metadados refinados:
- **`banca`**: Competição normalizada (ex: `OBM`, `IMO`, `HMMT`, `Baltic Way`)
- **`ano`**: Ano resolvido (ex: `2002`)
- **`edicao`**: Edição da competição (ex: `XXIV`, `November`, `1st`)
- **`etapa`**: Fase ou etapa identificada (ex: `Final Round`, `Nível 2`, `Round 2`)
- **`tipo_material`**: Identificação estrutural do material (`competicao`, `shortlist`, `seletiva`, `treinamento`)
- **`pais`**: País de origem da competição
- **`topico`**: Tópico matemático mapeado para as categorias canônicas do LEMMAS
- **`enunciado`**: Enunciado reconstruído em PT-BR (com integridade estrutural e fórmulas LaTeX)
- **`gabarito` / `solucao`**: Gabarito e resolução passo a passo (`solutions_markdown`)
- **`mathnet_id`**: Identificador original que garante idempotência (upsert sem duplicar)

```python
# ==============================================================================
# 8.1. MAPEAMENTO E PREPARAÇÃO DOS REGISTROS PARA O SUPABASE
# ==============================================================================
import sys
import importlib
from pathlib import Path
import json
import numpy as np
import pandas as pd

# Garante acesso aos módulos src.database.db e scripts.ingest do LEMMAS
NOTEBOOK_DIR = Path().resolve()
PROJECT_ROOT = NOTEBOOK_DIR.parent.parent if NOTEBOOK_DIR.name == 'notebooks' else NOTEBOOK_DIR.parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

# Recarrega o módulo em memória para o kernel vivo do Jupyter reconhecer as atualizações
import src.database.db
importlib.reload(src.database.db)
if 'db_supabase' in sys.modules:
    import db_supabase
    importlib.reload(db_supabase)

from src.database.db import get_supabase, inserir_questoes_lote
from scripts.ingest.mathnet_taxonomy import traduzir_taxonomia_mathnet

def extrair_str_segura(v):
    """Converte com segurança valores nulos, arrays numpy e strings vazias para None ou texto limpo."""
    if v is None:
        return None
    if isinstance(v, (list, np.ndarray)):
        return ', '.join(str(x) for x in v) if len(v) > 0 else None
    if pd.isna(v):
        return None
    s = str(v).strip()
    return None if s.lower() in ('<na>', 'nan', 'none', '') else s

# 2. Preenche tipo_material padrão como 'competicao' caso não seja shortlist/seletiva/treinamento
df['tipo_material_final'] = df['tipo_material'].fillna('competicao')

# 3. Filtra questões elegíveis:
#    - Possui enunciado traduzido / validado
#    - Pertence a bancas consolidadas (> 50 questões no banco)
bancas_top = df['banca'].value_counts()[lambda x: x > 50].index.tolist()
mascara_elegiveis = (
    df['enunciado_ptbr_reconstruido'].notna() &
    df['banca'].isin(bancas_top)
)

df_pronto_banco = df[mascara_elegiveis].copy()

print(f'Total de questões estruturadas prontas para o Supabase: {len(df_pronto_banco)}')
print(f'Bancas contempladas (> 50 questões): {len(bancas_top)}')
print('\nDistribuição por Tipo de Material:')
print(df_pronto_banco['tipo_material_final'].value_counts())

```

```python
# ==============================================================================
# 8.2. CONVERSÃO PARA O FORMATO CANÔNICO DO LEMMAS (TAXONOMIA EM PT-BR)
# ==============================================================================

registros_para_inserir = []

for _, row in df_pronto_banco.iterrows():
    # Enunciado em português com fallback
    enunciado_final = extrair_str_segura(row.get('enunciado_ptbr_reconstruido'))
    if not enunciado_final:
        enunciado_final = extrair_str_segura(row.get('enunciado')) or ''
        
    ano_val = int(row['ano']) if pd.notna(row.get('ano')) else None
    gabarito_val = extrair_str_segura(row.get('gabarito')) or 'Discursiva'
    solucao_val = extrair_str_segura(row.get('solucao_markdown'))
    
    # Tradução exaustiva da taxonomia do MathNet para PT-BR
    materia_pt, topico_pt, subtopico_pt = traduzir_taxonomia_mathnet(row.get('topico'))

    registro = {
        'materia': materia_pt,
        'topico': topico_pt,
        'subtopico': subtopico_pt,
        'dificuldade': 5,  # Nível Olímpico / Avançado
        'banca': str(row['banca']),
        'ano': ano_val,
        'edicao': extrair_str_segura(row.get('edicao_competicao')),
        'etapa': extrair_str_segura(row.get('etapa')),
        'tipo_material': str(row['tipo_material_final']),
        'pais': extrair_str_segura(row.get('pais')),
        'enunciado': enunciado_final,
        'gabarito': gabarito_val,
        'solucao': solucao_val,
        'figura_path': None,
        'tipo': 'discursiva',
        'mathnet_id': str(row['id']),
        'estrategias_esperadas': [topico_pt, 'Demonstração Axiomática'],
    }
    registros_para_inserir.append(registro)

print(f'Total de registros formatados: {len(registros_para_inserir)}')
print('\nExemplo de Registro Formatado com Taxonomia Canônica:')
import pprint
pprint.pprint(registros_para_inserir[0])

```

```python
# ==============================================================================
# 8.3. INGESTÃO NO SUPABASE (LOTE CONFIGURÁVEL COM UPSERT IDEMPOTENTE)
# ==============================================================================
# Defina a quantidade de questões a inserir agora (ex: 200 de teste, ou todas as 18.000)
LOTE_INICIAL = 200  # Ajuste para len(registros_para_inserir) para importar todo o acervo

# Prioriza as questões da OBM (Brasil) e competições internacionais clássicas (IMO, Baltic Way, HMMT)
def criterio_prioridade(r):
    b = r['banca']
    if b == 'OBM': return 0
    if b == 'IMO': return 1
    if b == 'Baltic Way': return 2
    if b == 'HMMT': return 3
    return 4

registros_priorizados = sorted(registros_para_inserir, key=criterio_prioridade)
lote_envio = registros_priorizados[:LOTE_INICIAL]

print(f'Iniciando envio de {len(lote_envio)} questões para o Supabase PostgreSQL...')
total_salvas = inserir_questoes_lote(lote_envio, chunk_size=50)
print(f'Sucesso! {total_salvas} questões inseridas/atualizadas com sucesso no Supabase!')
```
