```python
# Instala a biblioteca enem se ainda não estiver instalada
#!pip install enem
import os
import sys
import json
import shutil
import subprocess
from pathlib import Path

# Garante acesso à raiz do projeto
BASE_DIR = Path.cwd().parent
sys.path.append(str(BASE_DIR))

from src.database.db import pegar_conexao
print("✅ Ambiente configurado para importação do ENEM!")
```

```python
def extrair_e_importar_enem(
    caminho_pdf_prova: str,
    caminho_pdf_gabarito: str | None = None,
    ano: int = 2023,
    pasta_saida: str = "../data/enem_extraido",
    questoes_apenas_matematica: bool = True
):
    """
    Executa a extração do ENEM via CLI do 'enem', processa o JSON gerado,
    copia as imagens para data/uploads/ e insere no banco de dados do MathAI.
    """
    pasta_saida_path = Path(pasta_saida)
    pasta_saida_path.mkdir(parents=True, exist_ok=True)
    pasta_uploads = BASE_DIR / "data" / "uploads"
    pasta_uploads.mkdir(parents=True, exist_ok=True)

    cmd = [sys.executable, "-m", "enem", str(caminho_pdf_prova), "-o", str(pasta_saida_path)]
    if caminho_pdf_gabarito:
        cmd.extend(["-g", str(caminho_pdf_gabarito)])

    print(f"🚀 Executando Enem Extractor para o ENEM {ano}...")
    resultado = subprocess.run(cmd, capture_output=True, text=True)
    if resultado.returncode != 0:
        print(f"❌ Erro na extração do ENEM:\n{resultado.stderr}")
        return 0

    print("✅ Extração do PDF concluída com sucesso! Processando questões...")

    # Localiza o arquivo JSON gerado na pasta de saída
    arquivos_json = list(pasta_saida_path.glob("*.json"))
    if not arquivos_json:
        print("⚠️ Nenhum arquivo JSON encontrado na pasta de saída.")
        return 0

    with open(arquivos_json[0], "r", encoding="utf-8") as f:
        dados_enem = json.load(f)

    con = pegar_conexao()
    cur = con.cursor()
    total_inseridas = 0

    questoes_lista = dados_enem if isinstance(dados_enem, list) else dados_enem.get("questions", [])

    for q in questoes_lista:
        num = q.get("number") or q.get("index") or 0
        # No 2º dia do ENEM (a partir de 2009), Matemática vai da questão 136 até a 180
        if questoes_apenas_matematica and not (136 <= int(num) <= 180):
            continue

        enunciado = q.get("statement", "") or q.get("enunciado", "") or q.get("text", "")
        alternativas = q.get("alternatives", {}) or q.get("options", {})
        
        texto_completo = enunciado.strip()
        if alternativas:
            texto_completo += "\n\n"
            for letra in ["A", "B", "C", "D", "E"]:
                if letra in alternativas:
                    texto_completo += f"({letra}) {alternativas[letra]}\n"

        gabarito = str(q.get("answer") or q.get("gabarito") or "").strip().upper()

        figura_relativa = None
        imagem_origem = q.get("image") or q.get("figure")
        if imagem_origem:
            caminho_img = pasta_saida_path / imagem_origem
            if caminho_img.exists():
                nome_destino = f"enem_{ano}_q{num}_{caminho_img.name}"
                destino = pasta_uploads / nome_destino
                shutil.copy2(caminho_img, destino)
                figura_relativa = f"data/uploads/{nome_destino}"

        try:
            cur.execute("""
                INSERT INTO questoes (
                    materia, topico, enunciado, gabarito, dificuldade,
                    banca, ano, figura_path, tipo
                ) VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?)
            """, (
                "Matemática", "ENEM", texto_completo, gabarito, 3,
                "ENEM", ano, figura_relativa, "objetiva"
            ))
            total_inseridas += 1
        except Exception as e:
            print(f"Erro ao inserir questão {num}: {e}")

    con.commit()
    print(f"🎉 ENEM {ano}: {total_inseridas} questões de Matemática importadas com sucesso!\n")
    return total_inseridas


def importar_enem_por_ano(
    ano: int,
    pasta_provas: str = "../data/provas_enem",
    pasta_saida: str = "../data/enem_extraido"
):
    """
    Localiza automaticamente os PDFs da prova e gabarito do ano especificado
    e executa a extração e importação.
    Exemplo de uso em loop:
        for ano in range(2009, 2025):
            importar_enem_por_ano(ano)
    """
    pasta = Path(pasta_provas)
    if not pasta.exists():
        pasta.mkdir(parents=True, exist_ok=True)
        print(f"📁 Pasta criada: {pasta.resolve()}. Coloque os PDFs lá dentro!")
        return 0

    # Procura arquivos PDF do ano especificado
    arquivos_ano = list(pasta.glob(f"*{ano}*.pdf"))
    if not arquivos_ano:
        print(f"⚠️ [ENEM {ano}] Nenhum PDF encontrado contendo '{ano}' em {pasta}. Pulando...")
        return 0

    # Identifica o gabarito e a prova
    gabarito_pdf = None
    prova_pdf = None

    for f in arquivos_ano:
        nome_lower = f.name.lower()
        if "gabarito" in nome_lower or "gab" in nome_lower or "key" in nome_lower:
            gabarito_pdf = str(f)
        elif "dia2" in nome_lower or "matematica" in nome_lower or "mat" in nome_lower or "caderno" in nome_lower or "prova" in nome_lower:
            prova_pdf = str(f)

    # Se não achou prova específica com dia2/mat, pega o primeiro que não for gabarito
    if not prova_pdf:
        candidatos = [str(f) for f in arquivos_ano if str(f) != gabarito_pdf]
        if candidatos:
            prova_pdf = candidatos[0]

    if not prova_pdf:
        print(f"⚠️ [ENEM {ano}] Não foi possível identificar o PDF da prova. Pulando...")
        return 0

    print(f"📄 [ENEM {ano}] Prova: {Path(prova_pdf).name} | Gabarito: {Path(gabarito_pdf).name if gabarito_pdf else 'Nenhum'}")
    return extrair_e_importar_enem(
        caminho_pdf_prova=prova_pdf,
        caminho_pdf_gabarito=gabarito_pdf,
        ano=ano,
        pasta_saida=f"{pasta_saida}/enem_{ano}"
    )
```

```python
import re
from langdetect import detect, DetectorFactory
from langdetect.lang_detect_exception import LangDetectException

def remover_latex(texto):
    if not isinstance(texto, str):
        return ""

    # Remove blocos $$ ... $$
    texto = re.sub(r"\$\$.*?\$\$", " ", texto, flags=re.DOTALL)

    # Remove expressões $ ... $
    texto = re.sub(r"\$.*?\$", " ", texto, flags=re.DOTALL)

    #Aqui tiramos as quebras de linha

    texto = re.sub(r"\s","",texto).strip()

    return texto
```

```python
def detectar_idioma_seguro(texto):
    if not texto or len(texto.strip()) < 3:
        return None
    try:
        return detect(texto)
    except LangDetectException:
        return None
    except Exception:
        return None

def processar_idioma(row):
    # 1. Se já veio preenchido no dataset, mantém
    if pd.notna(row["idioma"]) and str(row["idioma"]).strip() not in ["", "nan", "NaN"]:
        return row["idioma"], "dataset"
    
    # 2. Se não veio, tenta detectar a partir do enunciado sem LaTeX
    idioma = detectar_idioma_seguro(row["enunciado_sem_latex"])
    if idioma is not None:
        return idioma, "detector"
    
    # 3. Casos que precisam de revisão manual ou IA posterior
    return None, "revisao"

# Aplica nas linhas criando idioma_final e origem_idioma
res = df.apply(processar_idioma, axis=1)
df["idioma_final"] = [r[0] for r in res]
df["origem_idioma"] = [r[1] for r in res]
```

```python
import re

MAPA_EXPANDIDO = {
    # Espanhol
    "spanish": "es", "español": "es", "espanol": "es", "castellano": "es",
    # Inglês
    "english": "en", "ingles": "en", "inglês": "en",
    # Português
    "portuguese": "pt", "português": "pt", "portugues": "pt",
    # Francês
    "french": "fr", "français": "fr", "francês": "fr",
    # Alemão
    "german": "de", "deutsch": "de",
    # Italiano
    "italian": "it", "italiano": "it",
    # Chinês
    "chinese": "zh", "chinese (traditional)": "zh", "chinese (simplified)": "zh", "mandarin": "zh",
    # Outros
    "russian": "ru", "mongolian": "mn", "romanian": "ro", "slovenian": "sl",
    "dutch": "nl", "korean": "ko", "turkish": "tr", "greek": "el",
    "ukrainian": "uk", "bulgarian": "bg", "croatian": "hr", "japanese": "ja",
    "czech": "cs", "hungarian": "hu", "polish": "pl", "vietnamese": "vi",
    "estonian": "et", "macedonian": "mk", "arabic": "ar", "hindi": "hi",
    "persian": "fa", "swedish": "sv", "finnish": "fi", "norwegian": "no",
    "catalan": "ca", "danish": "da"
}

def corrigir_idioma(row):
    # Pega o que está em idioma_limpo atualmente
    valor = str(row["idioma_limpo"]).strip().lower() if pd.notna(row["idioma_limpo"]) else ""
    
    # 1. Se tem múltiplos nomes com vírgula ou barra (ex: "english, arabic", "macedonian, english")
    if re.search(r"[,;/]| and ", valor):
        detectado = detectar_idioma_seguro(row["enunciado_sem_latex"])
        if detectado:
            return detectado.split("-")[0]
            
    # 2. Se for um nome por extenso (ex: "español" -> "es", "chinese (traditional)" -> "zh")
    if valor in MAPA_EXPANDIDO:
        return MAPA_EXPANDIDO[valor]
        
    # 3. Se já for código de 2 letras limpo (ex: "en", "pt", "es")
    if len(valor) == 2 and valor.isalpha():
        return valor
        
    # 4. Se tiver hífen (ex: "zh-cn" -> "zh")
    if "-" in valor:
        return valor.split("-")[0]
        
    # 5. Se nada acima resolveu, tenta o detector no texto
    detectado = detectar_idioma_seguro(row["enunciado_sem_latex"])
    if detectado:
        return detectado.split("-")[0]
        
    return None

# Atualiza diretamente a coluna 'idioma_limpo' e 'idioma'
df["idioma_limpo"] = df.apply(corrigir_idioma, axis=1)
df["idioma"] = df["idioma_limpo"]

# Exibe o resultado da própria 'idioma_limpo'
print("🌍 Ranking Final de idioma_limpo:")
print(df["idioma_limpo"].value_counts(dropna=False).head(20))
```

```python
ROMANO_PARA_ANO_OBM = {
    "I": 1979, "II": 1980, "III": 1981, "IV": 1982, "V": 1983,
    "VI": 1984, "VII": 1985, "VIII": 1986, "IX": 1987, "X": 1988,
    "XI": 1989, "XII": 1990, "XIII": 1991, "XIV": 1992, "XV": 1993,
    "XVI": 1994, "XVII": 1995, "XVIII": 1996, "XIX": 1997, "XX": 1998,
    "XXI": 1999, "XXII": 2000, "XXIII": 2001, "XXIV": 2002, "XXV": 2003,
    "XXVI": 2004, "XXVII": 2005, "XXVIII": 2006, "XXIX": 2007, "XXX": 2008,
    "XXXI": 2009, "XXXII": 2010, "XXXIII": 2011, "XXXIV": 2012, "XXXV": 2013,
    "XXXVI": 2014, "XXXVII": 2015, "XXXVIII": 2016, "XXXIX": 2017, "XL": 2018
}


# Pega o número romano antes de OBM (ex: "XXIV OBM", "XXXI Brazilian Math Olympiad")
regex_romano = r"\b([IVXLCDM]+)\s*(?:OBM|Brazilian Math Olympiad)\b"

# Extrai apenas a letra romana (ex: "XXIV", "III", "XXXI")
edicao_romana = df["banca"].str.extract(regex_romano, flags=re.IGNORECASE)[0]




```
